```python
plt.hist (df['units sold'], bins=6,edgecolor = "red", alpha = 0.9)

plt.title("Sales Distribution")

plt.xlabel("Sales")

plt.ylabel("Frequency")


plt.show()
```

```python
grouped_df = df.groupby('Publishing Year')['units sold'].sum().sort_values(ascending=False).head(15)

plt.figure(figsize=(12, 6))
plt.bar(grouped_df.index.astype(str), grouped_df.values)

plt.title('Top 15 Publishing Years by Units Sold')
plt.xlabel('Publishing Year')
plt.ylabel('Total Units Sold')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

the most units are sold in the year 2012 

```python
grouped_df = df.groupby('Author')['Book_average_rating'].sum().sort_values(ascending=False).head(5)

plt.figure(figsize=(8, 8))
plt.pie(
    grouped_df,
    labels=grouped_df.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 Authors by Total Book Rating')
plt.show()

print(grouped_df)
```

the top author by the book ranking is Stephen King 

```python
df = df[df["Publishing Year"]>1900]
```

```python
df.describe()
```

```python
df.isna().sum()
```

```python
df.dropna(subset="Book Name",inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.nunique()
```

```python
sns.histplot(df['Publishing Year'], kde=True, color='red')
plt.title('Distribution of year')
plt.ylabel('Frequency')
plt.show() 
```

after2000 we have seen growth in the publishing of books 

```python
grouped_df = df.groupby('Book Name')['gross sales'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))
plt.barh(grouped_df.index, grouped_df.values)

plt.title('Top 10 Books by Gross Sales')
plt.xlabel('Gross Sales')
plt.ylabel('Book Name')

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()

print(grouped_df)
```

the most gross sales book had is Go set a Watchman

```python
sns.lineplot(x='Author_Rating', y='sales rank', data=df)

plt.xticks (rotation=45)

plt.title("Sales Trend")

plt.show()
```

it shows the most sales rank as excellent

```python
#Bar Plot (Region-wise Sales)

sns.barplot(x='language_code', y='sales rank', data=df)

plt.title("Sales by Region")

plt.show() 
```

above shows the sales rank and language code 

```python
top_authors = df['Author'].value_counts().head(10) color='red'

plt.figure(figsize=(10,5))
top_authors.plot(kind='bar')
plt.title('Top 10 Authors by Number of Books')
plt.xlabel('Author')
plt.ylabel('Number of Books')
plt.xticks(rotation=45)
plt.show()
```

Top 10 Authors by Number of Books

```python
plt.figure(figsize=(9,6))
plt.scatter(df['Book_average_rating'], df['Book_ratings_count'], alpha=0.5)
plt.title('Book Average Rating vs Number of Ratings')
plt.xlabel('Book Average Rating')
plt.ylabel('Book Ratings Count')
plt.show()
```

Book Rating vs Ratings Count

```python
avg_rating = df.groupby('genre')['Book_average_rating'].mean().sort_values(ascending=False)

plt.figure(figsize=(9,5))
avg_rating.plot(kind='bar')
plt.title('Average Book Rating by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.show()
```

Average Rating by Genre
